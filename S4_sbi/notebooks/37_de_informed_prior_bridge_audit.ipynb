## Handoff inspection

Load `prior_specification.json` with fail-closed acknowledgment of restrictions
(Notebook 38 pattern).

```python
from pathlib import Path
import json

ROOT = Path(r"D:/Year3_Mao_Projects/sleep_loop")
spec_path = Path(result["prior_specification_path"])
if not spec_path.is_absolute():
    spec_path = ROOT / spec_path
gate_path = Path(result["decision_gate_path"])
if not gate_path.is_absolute():
    gate_path = ROOT / gate_path

gate = json.loads(gate_path.read_text(encoding="utf-8"))
spec = json.loads(spec_path.read_text(encoding="utf-8"))

print("claim_route:", result["claim_route"])
print("statistical_role:", result["statistical_role"])
print("track_outcome:", result["track_outcome"])
print("primary_track:", result["primary_track"])
print("c_ctx2th_verdict:", result["c_ctx2th_verdict"])
print("decision_status:", result["decision_status"])
print("personalized_twin_bridge_status:", result["personalized_twin_bridge_status"])
print("fallback_chapter_status:", result["fallback_chapter_status"])
print("estimator_handoff_branch:", result["estimator_handoff_branch_eligible"])
print("no_go_path:", gate.get("no_go_path"))
print("n_restrictions:", len(result["restrictions"]))

if spec["decision_status"] in {"GO", "CONDITIONAL_GO"}:
    loaded = load_prior_specification(
        spec_path,
        purpose="personalized_twin_inference",
        required_claim_route=spec["claim_route"],
        required_role=spec["statistical_role"],
        required_track=spec["primary_track"],
        acknowledge_restrictions=list(spec.get("restrictions") or []),
    )
    print("loader_ok:", loaded["decision_status"])
else:
    print("loader_skipped_due_to_decision_status:", spec["decision_status"])

fig_dir = Path(result["artifact_root"])
if not fig_dir.is_absolute():
    fig_dir = ROOT / fig_dir
fig_dir = fig_dir / "figures"
pngs = sorted(p.name for p in fig_dir.glob("*.png"))
assert len(pngs) == 4, pngs
print("figures:", pngs)
```
