```python
import json, sys
from pathlib import Path

import numpy as np
import pandas as pd

SRC = Path.cwd().resolve().parents[0] / "src"
if not SRC.is_dir():
    SRC = Path.cwd().resolve().parents[1] / "S4_sbi" / "src"
if str(SRC) not in sys.path:
    sys.path.insert(0, str(SRC))

from sleep_sbi import figure10_closure_claude_common as common
from sleep_sbi.figure10_protocol import PROJECT_ROOT

pd.set_option("display.width", 160)
pd.set_option("display.max_columns", 40)


def show(payload, keys=None):
    """Print a JSON-ish payload compactly, optionally restricted to some keys."""
    if keys is not None:
        payload = {k: payload[k] for k in keys if k in payload}
    print(json.dumps(payload, indent=2, ensure_ascii=False, default=str)[:4000])


print("python  :", sys.version.split()[0])
print("env     :", Path(sys.executable).parent.name)
print("src     :", SRC)
```

```python

from sleep_sbi import figure10_notebook41_claude as nb41

paths = common.claude_dirs(41)
common.assert_immutables()
handoff39, nb40_claude = nb41.load_handoffs()
print("39 -> 40 handoff:", handoff39["claude40_handoff_status"])
print("40 -> 41 handoff:", nb40_claude["claude41_handoff_status"])
print("7D direction    :", nb40_claude["seven_d"]["descriptive_evidence_direction"])
print("8D direction    :", nb40_claude["eight_d"]["descriptive_evidence_direction"])
```


## Phase 1 — Intervention justification

`CONTROL_NOT_JUSTIFIED` is the right call and the Cursor closure reached it. What it did
not do is *demonstrate* it — the status was frozen by assertion before any artifact was
consulted. Here each of the four elements a preregistered intervention needs is searched
for, by name, across every lock and preregistration in the project.


```python

intervention = nb41.phase1_intervention(paths)
display(pd.read_csv(paths["csv"] / "intervention_elements.csv"))
print("artifacts searched:")
for artifact in intervention["artifacts_searched"]:
    print("   ", artifact)
print("missing:", intervention["missing_artifacts"] or "(none)")
print()
print(intervention["argument"])
print()
print("STATUS:", intervention["intervention_status"],
      "| simulations:", intervention["intervention_simulations_run"])
```


## Phase 2 — Decision sensitivity

Five readings of the same evidence. One of them is the shipped closure; the other four are
the readings that survive the Notebook 39 admissibility gate.


```python

sensitivity = nb41.phase2_decision_sensitivity(paths, nb40_claude=nb40_claude,
                                               handoff39=handoff39)
display(pd.read_csv(paths["csv"] / "decision_sensitivity.csv"))
```

```python

for reading in sensitivity["readings"]:
    mark = "admissible" if reading["admissible"] else "INADMISSIBLE"
    print("─" * 100)
    print(f"{reading['reading']}  [{mark}]")
    print("   families :", ", ".join(reading["families_counted"]) or "(none)")
    print("   7D       :", reading["direction_7d"], "->", reading["terminal_status"])
    print("   why      :", reading["why"])
```

```python

print("admissible terminal statuses:", sensitivity["admissible_terminal_statuses"])
print("invariant across them       :",
      sensitivity["terminal_status_invariant_across_admissible_readings"])
print("shipped status              :", sensitivity["shipped_status"])
print("shipped is the only outlier :", sensitivity["shipped_status_is_the_only_outlier"])
print()
alpha = sensitivity["alpha_independence"]
print(alpha["claim"])
print(alpha["reason"])
print("   candidates beating V7        : %d/%d" % (alpha["n_better_than_v7_observed"],
                                                   alpha["n_candidates"]))
print("   needed for significant FAV   : %d" % alpha["n_better_required_for_significant_favorable"])
print("   shortfall                    : %d" % alpha["shortfall"])
print()
print(sensitivity["single_metric_dependence"])
```


## Phase 3 — Terminal status

The frozen 7D-only mapping, applied to the admissible evidence. 8D contributes
`sensitivity_concordance` and nothing else.


```python

terminal = nb41.phase3_terminal(paths, nb40_claude=nb40_claude, handoff39=handoff39,
                                sensitivity=sensitivity, intervention=intervention)
show(terminal, ["global_uncertainty_status", "predictive_twin_status",
                "descriptive_evidence_direction_7d", "descriptive_evidence_direction_8d",
                "sensitivity_concordance", "sensitivity_concordance_is_deciding",
                "intervention_status", "final_part_ii_status", "decided_by",
                "invariant_across_admissible_readings", "admissible_families",
                "excluded_families", "failure_reasons"])
```


## Phase 4 — The dissertation paragraph

Every clause traced to the artifact that supports it, and the shipped sentence retracted
with its counter-evidence attached.


```python

paragraph = nb41.phase4_paragraph(paths, terminal=terminal, nb40_claude=nb40_claude,
                                  sensitivity=sensitivity)
for i, clause in enumerate(paragraph["clauses"], start=1):
    print("─" * 100)
    print(f"[{i}] {clause['clause']}")
    print(f"     evidence: {clause['evidence']}")
```

```python

print("PARAGRAPH\n")
print(paragraph["dissertation_paragraph"])
print()
print("=" * 100)
print("RETRACTED\n")
print(paragraph["retraction"]["retracted_sentence"])
print()
print(paragraph["retraction"]["reason"])
```


## Phase 5 — Final audits


```python

audits = nb41.phase5_final_audits(paths)
show(audits, ["sealed", "no_notebook_42", "write_scope", "environment",
              "locks_read_not_written", "integrity_status"])
assert audits["integrity_status"] == "INTEGRITY_OK"
print("\nartifacts hashed:", len(audits["artifact_hashes"]))
```
